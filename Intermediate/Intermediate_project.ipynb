```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))
plt.plot(df["date"], df["pm2_5"])
plt.title("PM2.5 Trend Over Time")
plt.xlabel("Date")
plt.ylabel("PM2.5")
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df["date"] = pd.to_datetime(df["date"])
df["Month"] = df["date"].dt.month_name()

plt.figure(figsize=(10,5))
df.groupby("Month")["pm2_5"].mean().plot(kind="bar", color="green")

plt.title("Average PM2.5 by Month")
plt.xlabel("Month")
plt.ylabel("Average PM2.5")
plt.grid(axis="y")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["pm2_5"], bins=20, color="orange", edgecolor="black")

plt.title("Distribution of PM2.5")
plt.xlabel("PM2.5")
plt.ylabel("Frequency")
plt.grid(axis="y")
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.boxplot(df["pm2_5"])

plt.title("Box Plot of PM2.5")
plt.ylabel("PM2.5")
plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(10,5))

df.groupby("Month")["pm2_5"].mean().plot(
    kind="line",
    marker="o",
    color="purple"
)

plt.title("Average PM2.5 by Month")
plt.xlabel("Month")
plt.ylabel("Average PM2.5")
plt.grid(True)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 8))

numerical_cols = ['pm2_5', 'pm10', 'no2', 'so2', 'co', 'o3', 'nh3']

correlation_matrix = df[numerical_cols].corr()

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title("Correlation Heatmap of Air Quality Parameters")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="pm2_5",
    y="pm10"
)

plt.title("PM2.5 vs PM10")
plt.xlabel("PM2.5")
plt.ylabel("PM10")
plt.grid(True)

plt.show()
```
